# Genie Dashboard Validation

#### Total Race Entries 
- Genie showed 6,669,244 total race entries which is correct

```python
%sql
SELECT
  *
FROM
  marathos.gold.v_time_top_events_by_finishers
ORDER BY
  most_finishers DESC
LIMIT 10;
```

#### Total rows
- Genie correct calculation

```python
%sql
SELECT
  COUNT(*)
FROM
  marathos.gold.fct_results;
```

#### Average speed of all runners
- Genie showed 7.24 - also correct

```python
%sql
SELECT
  AVG(CAST(athlete_average_speed AS DOUBLE)) AS avg_speed_kmh
FROM
  marathos.gold.fct_results;
```

#### Total countries
- Genie was correct with 198

```python
%sql
SELECT
  COUNT(DISTINCT a.athlete_country)
FROM
  marathos.gold.fct_results f JOIN marathos.gold.dim_athlete a USING (athlete_id)
WHERE
  a.athlete_country IS NOT NULL;
```

#### Countries with most participants
- Genie was correct with the top 5 countries: USA, FRA, RSA, GER, ITA

```python
%sql
SELECT
  a.athlete_country,
  COUNT(*) AS entries
FROM
  marathos.gold.fct_results f JOIN marathos.gold.dim_athlete a USING (athlete_id)
GROUP BY
  a.athlete_country
ORDER BY
  entries DESC
LIMIT 5;
```

#### Years with most runners
- Peak: 2019 with 715,562 entries. 2020 dip to 219,523 (-69%). Recovery to 459K in 2022.
- Matches manual count from fct_results.

```python
%sql
SELECT
  year_of_event,
  COUNT(*) AS entries
FROM
  marathos.gold.fct_results
GROUP BY
  year_of_event
ORDER BY
  entries DESC
LIMIT 5;
```

#### Gender ratio
- Genie showed 79,92% male and 20,08% female which is correct

```python
%sql
SELECT
  athlete_gender,
  COUNT(*) AS entries
FROM
  marathos.gold.fct_results f JOIN marathos.gold.dim_athlete a USING (athlete_id)
GROUP BY
  athlete_gender
ORDER BY
  entries DESC;
```
