# Genie Dashboard Validation

#### Total Race Entries 
- Genie showed 342,85 mn (summed event_number_of_finishers instead of counting rows)

```python
%sql
SELECT
  *
FROM
  marathos.gold.v_time_top_events_by_finishers
ORDER BY
  most_finishers DESC
LIMIT 10;
```

#### Total rows
- Genie correct calculation

```python
%sql
SELECT
  COUNT(*)
FROM
  marathos.gold.fct_results;
```

#### Average speed of all runners
- Genie showed 7.89 km/h instead of 7.24 (probably cache problem, same SQL gave different answers)

```python
%sql
SELECT
  AVG(CAST(athlete_average_speed AS DOUBLE)) AS avg_speed_kmh
FROM
  marathos.gold.fct_results;
```

#### Total countries
- Genie were correct

```python
%sql
SELECT
  COUNT(DISTINCT a.athlete_country)
FROM
  marathos.gold.fct_results f JOIN marathos.gold.dim_athlete a USING (athlete_id)
WHERE
  a.athlete_country IS NOT NULL;
```

#### Countries with most participants

- Genie showed TPE at the top instead of USA/FRA (summed finishers, not rows)

```python
%sql
SELECT
  a.athlete_country,
  COUNT(*) AS entries
FROM
  marathos.gold.fct_results f JOIN marathos.gold.dim_athlete a USING (athlete_id)
GROUP BY
  a.athlete_country
ORDER BY
  entries DESC
LIMIT 5;
```

#### Years with most runners
- Genie showed "mn" values ​​(summed finishers)

```python
%sql
SELECT
  year_of_event,
  COUNT(*) AS entries
FROM
  marathos.gold.fct_results
GROUP BY
  year_of_event
ORDER BY
  entries DESC
LIMIT 5;
```

#### Gender ratio
- Genie showed 286.81 million for men (same problem as Total race entries)

```python
%sql
SELECT
  athlete_gender,
  COUNT(*) AS entries
FROM
  marathos.gold.fct_results f JOIN marathos.gold.dim_athlete a USING (athlete_id)
GROUP BY
  athlete_gender
ORDER BY
  entries DESC;
```
